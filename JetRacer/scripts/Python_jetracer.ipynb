## 1.2 Keep a value under a name

A **variable** is a name stuck onto a value, so you can use it again.

`throttle = 0.3` stores `0.3` under the name `throttle`.
The name is on the left. The value is on the right.
`=` means **store**. It is not the equals sign from maths.

On this car, throttle is from `0` (stop) to `1` (full power).

````python
throttle = 0.3
print(throttle)
````

Write the name again and Python fetches the stored value.
Store a new value under the same name and the old one is replaced.

A comma inside `print` puts a space between the parts.

````python
throttle = 0.0
print("stop, throttle =", throttle)
````

## 1.4 Maths on Jetracer

`+` `-` `*` `/` mean add, subtract, multiply, divide.
`*` and `/` happen before `+` and `-`, same as school maths.
Anything inside `( )` happens first. If the order is not obvious, add brackets.

````python
# the sensor reports millimetres; we want centimetres
# structure: variable = math operation
distance_mm = 450
distance_cm = distance_mm / 10
print(distance_cm, "cm")
````

````python
# two speed checks, then their average
speed_mps = (1.1 + 1.5) / 2
print(speed_mps, "m/s")
````

`//` divides and **throws away** the fraction. That is how many whole minutes fit.
`%` is the **leftover**. That is the extra seconds.

100 seconds is 1 minute and 40 seconds:
`100 // 60` is `1`, and `100 % 60` is `40`.

````python
seconds = 100
minutes = seconds // 60
left = seconds % 60
print(minutes, "min", left, "sec")
````

# 2. Control Structures

The JetRacer should not always do the same thing.
If a wall is close, stop. While the road is clear, keep going.

Today the car still stays still. You practise `if` and `while`.
Press **Shift+Enter** to run a cell.

## 2.1 A yes-or-no question

A **condition** is a question. Python answers `True` or `False`.

| Sign | Means |
| --- | --- |
| `<` `<=` | smaller, smaller or same |
| `>` `>=` | bigger, bigger or same |
| `==` | same value |
| `!=` | not the same |

`=` stores a value. `==` asks "are these the same?".
One equals sign inside an `if` is a mistake.

````python
distance_mm = 150
print(distance_mm < 200)   # True, the wall is close
print(distance_mm == 150)  # True, same number
print(distance_mm != 150)  # False
````

## 2.2 `if` — do this only when the answer is yes

Ask the question. `true` takes the extra step. `false` skips it.

![Flowchart: Branching](https://i.cs.hku.hk/~yklam2/MScPy/img/1-2/flow-chart-branch.svg)

In Python that extra step is the indented block under `if`. The line with `if` ends in a colon `:`.

![Flowchart: If-then](https://i.cs.hku.hk/~yklam2/MScPy/img/1-2/flow-chart-if-then.svg)

Lines pushed in by 4 spaces belong to the `if`. A line that starts at the left always runs.

````python
distance_mm = 150
if distance_mm < 200:
    print("stop")
print("check done")
````

Change the distance to 450 and run again. `450 < 200` is `False`, so `"stop"` is skipped. `"check done"` still prints, because it is not indented.

````python
distance_mm = 450
if distance_mm < 200:
    print("stop")
print("check done")
````

## 2.3 `else` — the other path

`true` runs the `if` block. `false` runs the `else` block. Then the two paths meet again.
Only one of the two blocks runs.

![Flowchart: If-then-else](https://i.cs.hku.hk/~yklam2/MScPy/img/1-2/flow-chart-if-then-else.svg)

````python
# structure: if-else
# if condition1:
#     do this
# else:
#     do this  # last option if condition1 is false

distance_mm = 450
if distance_mm < 200:
    throttle = 0.0
else:
    throttle = 0.3
    print("go")
print("throttle", throttle)
````

## 2.4 `elif` — more than two answers

Steering is negative for left, zero for straight, positive for right.
Python checks the questions from the top and **stops at the first `True`**. The other blocks are skipped.

![Flowchart: If-then-elif](https://i.cs.hku.hk/~yklam2/MScPy/img/1-2/flow-chart-if-then-elif.svg)

````python
# structure: if-elif-else
# if condition1:
#     do this
# elif condition2:
#     do this
# else:
#     do this  # last option if all above are false

steering = -0.6
if steering < 0:
    print("left")
elif steering > 0:
    print("right")
else:
    print("straight")
````

## 2.5 Two questions at once

Yes means `True`. No means `False`.

| Question A | Question B | `and` | `or` |
| --- | --- | --- | --- |
| yes | yes | yes | yes |
| yes | no | no | yes |
| no | yes | no | yes |
| no | no | no | no |

`and` needs **both** yes. `or` needs **one** yes.
`not` flips the answer: yes becomes no, no becomes yes.

````python
distance_mm = 150
throttle = 0.0
# structure: condition1 and/or condition2  --> combine to produce final answer
print(distance_mm < 200 and throttle > 0)  # False: close, but already stopped
print(distance_mm < 200 or throttle > 0)    # True: the wall is close
print(not throttle > 0)                     # True: we are not moving
````

## 2.6 `while` — repeat while the answer stays yes

Check the condition. `true` does the step, then the arrow goes **back** and checks again.
`false` leaves the dashed box and the program ends.
The dashed box is the loop.

![Flowchart: Looping](https://i.cs.hku.hk/~yklam2/MScPy/img/1-3/flow-chart-loop.svg)

The indented lines are the step.
If a cell never finishes, click the **stop** button (the square) in the toolbar.

`seconds = seconds - 1` can also be written `seconds -= 1`. Same thing.

````python
seconds = 3
while seconds > 0:
    print(seconds)
    seconds = seconds - 1
print("go")
````

# 3. Functions

A **function** is a named recipe. You write the steps once, then run them by name.
`print` and `float` are functions you already use. Today you write your own.

The car still stays still. Press **Shift+Enter** to run a cell.

## 3.1 Name the steps

A function is a helper. Write the steps once, give them a name, and reuse that name. The name is easy to recall, and the steps sit in one place, so they stay clear and easy to manage.

The program jumps into the function, runs the indented lines, then comes back.

![Flowchart: Function](https://i.cs.hku.hk/~yklam2/MScPy/img/1-5/flow-chart-function.svg)

`def` starts the self-deined function. The line ends with `:`. The indented lines are the steps.
Writing `def` does not run the steps. The name with `()` does.

````python
def say_ready():
    print("JetRacer ready")

say_ready()
say_ready()
````

## 3.2 Pass a value in

The name in the parentheses is filled by the call.
`show_throttle(0.3)` stores `0.3` in `throttle` for that run only.

````python
def show_throttle(throttle):
    print("throttle", throttle)

show_throttle(0.3)
show_throttle(0.0)
````

The function gets its own copy of the number. The `throttle` above the function stays `0.2`.

````python
# global variable
throttle = 0.2

def show_throttle(throttle):
    # local variable
    throttle = 0.0
    print("inside", throttle)

show_throttle(throttle)
print("outside", throttle)
````

## 3.3 Hand a number back

`return` sends a value back to the call, and the function stops there.
You can store that value. A `print` inside the function only shows it.

Throttle on this car must stay from `-1` to `1`.

````python
def limit_throttle(throttle):
    if throttle > 1:
        return 1
    if throttle < -1:
        return 0
    return throttle

print(limit_throttle(0.4))
print(limit_throttle(1.5))
print(limit_throttle(-1.2))
````

A function can take more than one value. Separate the names with a comma.
Speed is distance divided by time, same as lab 1.

````python
def speed_mps(distance_m, seconds):
    return distance_m / seconds

print(speed_mps(1.5, 3))
````
